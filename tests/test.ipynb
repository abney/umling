```python
from umling.mol import *
```

```python
Letter = crange('a', 'z') + crange('A', 'Z')
```

```python
Digit = alphabet('0123456789')
```

```python
Underscore = sym('_')
```

```python
Alphanumeric = Letter + Underscore + Digit
```

```python
Identifier = (Letter + Underscore) * star(Alphanumeric)
```

```python
LBracket = sym('[')
```

```python
RBracket = sym(']')
```

```python
Arrow = sym('-') * sym('>')
```

```python
Newline = opt(sym('\r')) * sym('\n')
```

```python
Whitespace = repeat(sym(' ') + sym('\t') + sym('\f') + sym('\v'))
```

```python
from umling.mol.symbols import ID, LB, RB, ARROW, NL, WS
```

```python
Token = (
    Identifier * io(epsilon, ID) +
    io(LBracket, LB) +
    io(RBracket, RB) +
    io(Arrow, ARROW) +
    io(Newline, NL) +
    io(Whitespace, WS)
)
```

```python
instring = letters('''
    S ->
    NP
''')
```

```python
from umling.mol.lang import Tokenizer
tkz = Tokenizer(Token)
```
